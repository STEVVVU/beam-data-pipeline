# Local Beam datamart and cloud SQL
Runs actual local CLI on the bundled fixtures. Cloud SQL is displayed for review only; no cloud resources are created.

```python
import subprocess
import sys
import tempfile
from pathlib import Path

run_dir = tempfile.mkdtemp(prefix='beam-notebook-')
for source in ('customers', 'products', 'orders'):
    subprocess.run([sys.executable, '-m', 'pipeline', '--source', source, '--input', f'fixtures/{source}.jsonl', '--output', str(Path(run_dir) / source)], check=True)
subprocess.run([sys.executable, '-m', 'reports', run_dir], check=True)
```
